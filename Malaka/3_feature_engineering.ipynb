```python
import pandas as pd, numpy as np, pickle, os
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
import scipy.sparse as sp
print("Ready!")
```

```python
df = pd.read_csv('cleaned_data.csv')
X = df['clean_text'].astype(str)
y = df['label']
print(f"Samples: {len(X):,}")
```

```python
bow = CountVectorizer(
max_features=10000,
ngram_range=(1, 2),
min_df=2,
max_df=0.95
)
X_bow = bow.fit_transform(X)
print(f"BoW Matrix: {X_bow.shape}")
print(f"Vocabulary size: {len(bow.vocabulary_):,}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X_bow, y, test_size=0.2,
random_state=42, stratify=y)
os.makedirs('saved_models', exist_ok=True)
pickle.dump(bow, open('saved_models/bow_vectorizer.pkl','wb'))
sp.save_npz('saved_models/X_train.npz', X_train)
sp.save_npz('saved_models/X_test.npz', X_test)
np.save('saved_models/y_train.npy', y_train)
np.save('saved_models/y_test.npy', y_test)
print("Saved! Train:", X_train.shape, "Test:", X_test.shape)
```
